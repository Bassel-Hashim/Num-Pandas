<span style="font-size:45px;">5 - Modifying Data</span>


```python
def inc10(x):
    return x*1.10

def inc1(x):
    return x+1
```

```python
df["Salary_Bonus"]=df["Salary"].apply(inc10)
df["Age"]=df["Age"].apply(inc10)
df = df.rename(columns={'Salary': 'Monthly_Salary'})
```

<span style="font-size:45px;">6 - Missing Data</span>


```python
import numpy as np
```

```python
df.loc[1, "Age"] = np.nan
df.loc[4, "City"] = None
```

```python
df["Age"].dtype
# pandas automatically upcasts the entire column's data type from int64 to float64 so it can seamlessly accommodate the floating-point NaN value.
```

```python
print(df.isnull().sum())
```

```python

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["City"] = df["City"].fillna("Unknown")
```

```python
df
```

```python
df.loc[1, "Age"] = np.nan
df.loc[4, "City"] = None

len_before = len(df)
print(f"Length before dropna: {len_before}")  # Output: 5

df_dropped = df.dropna()
len_after = len(df_dropped)
print(f"Length after dropna: {len_after}")  # Output: 3
```

```python
avg_salary = df.groupby("Department")["Monthly_Salary"].mean()
print(avg_salary)
```

```python
summary_stats = df.groupby("Department")["Monthly_Salary"].agg(
    ["mean", "max", "count"]
)
print(summary_stats)
```

```python
dept_counts = df["Department"].value_counts()
print(dept_counts)
```

```python
df["Level"] = df["Monthly_Salary"].apply(
    lambda x: "High" if x >= 15000 else "Low"
)
print(df)
```

```python
top_3_earners = df.sort_values(by="Monthly_Salary", ascending=False).head(3)
print(top_3_earners)
```
