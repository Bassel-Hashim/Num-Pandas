<span style="font-size:45px;"><span style="font-size:45px;">3. Concatenation and the "danger zone"</span>
</span>


```python
x = np.ones((2, 3)); y = np.zeros((2, 3))
print(np.concatenate([x, y], axis=0).shape)   # (4, 3)
print(np.concatenate([x, y], axis=1).shape)   # (2, 6)

try:
    np.concatenate([np.ones((2, 3)), np.ones((2, 2))], axis=0)
except ValueError as e:
    print("Error:", e)
```

<span style="font-size:45px;">5. Views and copies</span>


```python
a = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float32)

row = a[0, :]
row[0] = 99
print(a)        
               
a = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float32)
row = a[0, :].copy()
row[0] = 99
print(a)        
```
