<span style="font-size:45px;">Merging and Joining</span>


```python
Dict1={"ID":[1,2,3,4],
       "Name":["Ali","Sara","Ahmed","Samy"],
       "Age":[20,25,30,35],
       "Salary":[1000,2000,3000,4000]}
df1=pd.DataFrame(Dict1)


Dict2={"ID":[1,2,3,1],
       "Department":["HR","PR","IT","HR"],
       }
df2=pd.DataFrame(Dict2)
```

```python
df1
```

```python
df2
```

```python
result= pd.merge(df1,df2,on="ID")
result
```

```python
result= pd.merge(df1,df2,on="ID", how="inner")
result
```

```python
result= pd.merge(df1,df2,on="ID", how="right")
result
```

```python
result= pd.merge(df1,df2,on="ID", how="left")
result
```

<span style="font-size:45px;">Concatenation</span>


```python
con= pd.concat([df1,df2])
con
```

```python
con= pd.concat([df1,df2], join='inner')
con
```

```python
con= pd.concat([df1,df2], join='inner', ignore_index=True)
con
```
